```python
from tensorflow.keras import datasets
import tensorflow.keras as tl

mist = datasets.mnist
(train_images, train_labels), (test_images, test_labels) = mist.load_data()

print(len(train_images))
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
unq_value , count = np.unique(train_labels, return_counts=True)

plt.pie(count, labels=unq_value , autopct='%1.1f%%')
plt.show()
```

```python
max = 25

image_4 = []

for i in range(len(train_images)):
  if train_labels[i] == 4:
    image_4.append(train_images[i])

    if len(image_4) == max:
      break

row = 5

col = 5

plt.figure(figsize=(10,10))

for i in range(max):
  plt.subplot(row, col, i+1)

  plt.xticks([])
  plt.yticks([])
  plt.imshow(image_4[i], cmap=plt.cm.berlin)

plt.show()
```

```python
train_images = train_images / 255.0
test_images = test_images / 255.0

model = tl.Sequential([
    tl.layers.Flatten(input_shape=(28,28)),
    tl.layers.Dense(512, activation='relu'),
    tl.layers.Dense(256, activation='relu'),
    tl.layers.Dense(128, activation='relu'),
    tl.layers.Dense(64, activation='relu'),
    tl.layers.Dense(32, activation='relu'),
    tl.layers.Dense(16, activation='relu'),
    tl.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(train_images , train_labels , epochs=10)
```

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print('test_acc' , test_acc)
```

```python
predictions = model.predict(test_images)

plt.figure(figsize=(10,10))
for i in range(25):
  plt.subplot(5,5,i+1)
  plt.xticks([])
  plt.yticks([])
  plt.grid(False)
  plt.imshow(test_images[i], cmap=plt.cm.berlin)

  predicted_label = np.argmax(predictions[i])
  true_label = test_labels[i]

  if predicted_label == true_label:
    color = 'blue'
  else:
    color = 'red'
  plt.xlabel("{}({})".format(predicted_label, true_label), color=color)

plt.show()
```
